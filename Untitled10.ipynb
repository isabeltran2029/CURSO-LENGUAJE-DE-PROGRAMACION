TRM DIA

trm_del_dia

Descripción

Obtiene la Tasa Representativa del Mercado (TRM) vigente para la fecha actual. La TRM se expresa en pesos colombianos (COP) por dólar estadounidense (USD) y la información proviene de la fuente oficial que publica este indicador.

Parámetros

No requiere parámetros.

Estructura de la respuesta

```python
{
  "fecha": "2026-10-24",
  "trm": 4210.80,
  "moneda_base": "USD",
  "moneda_destino": "COP",
  "unidad": "COP por USD"
}
```

Cuando no hay datos

```python
{
  "fecha": "2026-10-24",
  "mensaje": "No hay TRM disponible para la fecha consultada."
}
```

Cuando falla la fuente

```python
error_fuente = {
  "error": True,
  "codigo": "TRM_NO_DISPONIBLE",
  "mensaje": "No fue posible obtener la TRM del día."
}
```

¿Por qué le sirve a alguien más?

Permite consultar rápidamente el valor oficial del dólar para realizar conversiones monetarias, pagos internacionales y análisis financieros

Ficha de la Herramienta 2
Nombre

trm_historica

Descripción

Obtiene la TRM para un rango de fechas específico. Los valores se expresan en pesos colombianos (COP) por dólar estadounidense (USD) y provienen de la fuente oficial de publicación de la TRM.

Parámetros
fecha_inicio (string, formato YYYY-MM-DD)
fecha_fin (string, formato YYYY-MM-DD)
incluir_variacion (boolean, valor por defecto: true)

```python
{
  "fecha_inicio": "2026-10-01",
  "fecha_fin": "2026-10-07",
  "trm_inicial": 4150.25,
  "trm_final": 4210.80,
  "variacion_porcentual": 1.46,
  "unidad": "COP por USD",
  "moneda_base": "USD",
  "moneda_destino": "COP"
}
```

Cuando no hay datos

```python
{
  "fecha_inicio": "2026-10-01",
  "fecha_fin": "2026-10-07",
  "mensaje": "No se encontraron datos de TRM para el rango solicitado."
}
```

Cuando falla la fuente

```python
{
  "error": True,
  "codigo": "TRM_HISTORICA_NO_DISPONIBLE",
  "mensaje": "No fue posible consultar la TRM histórica."
}
```

¿Por qué le sirve a alguien más?

Permite analizar la evolución del dólar en un período determinado, facilitando la toma de decisiones financieras, comerciales y de inversión.

Conclusión

Las herramientas trm_del_dia y trm_historica fueron diseñadas para resolver necesidades diferentes: una entrega un valor puntual y la otra permite analizar tendencias en un rango de fechas. Además, ambas devuelven respuestas estructuradas que facilitan la interpretación y el uso correcto de la información por parte de un agente MCP.
